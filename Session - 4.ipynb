```python
# Question 1

products = ["Pen", "Notebook", "Marker", "Folder", "Stapler"]

first_product = products[0]
last_products = products[-1]
middle_products = products[1:4]
reversed_products = products[::-1]

print("Products:", products)
print("First:", first_product)
print("Last:", last_products)
print("Middle:", middle_products)
print("Reversed:" ,reversed_products)
```

```python
# Question 2

items = ["Pen", "Notebook"]
new_items = ["Marker", "Folder"]

appended = items.copy()
appended.append(new_items) # [pen, notebook, [marker, folder]]

extended = items.copy()
extended.extend(new_items) # [pen, notebook, marker, folder]

print("Original:", items)
print("Appended:", appended)
print("Extended:", extended)
```

```python
# Question 3

cities = ["Delhi", "Mumbai", "Pune", "Jaipur"]
removed_city = "Pune" 

print("Starting:", cities)

cities.insert(1, "Chennai")
print("After insertion:", cities)

cities.remove(removed_city)
print("After removal:", cities)

last_city = cities.pop()

print("After pop:", cities)
print("Popped city:", last_city)
```

```python
# Question 4

scores = [72, 91, 68, 85, 77]

sorted_scores = sorted(scores)

print("Original:", scores)
print("Sorted copy:", sorted_scores)

scores.sort(reverse=True)

print("Original after sort:", scores)
print("Sorted copy:", sorted_scores)
```

```python
# Question 6

original = [10, 20, 30]

alias = original 
alias.append(40)

print("Original:", original)
print("Alias:", alias)
print("Same list:", alias is original) #True
```

```python
# Question 7

original = [[1], [2]]

independent = [original[0].copy(), original[1].copy()]

original[0].append(9)

print("Original:", original)
print("Independent:", independent)
```
