```python
# Question 1

product = {'id': 101, 'name': 'Notebook', 'price': 125.0}

print("Product:", product)
print("ID:", product['id'])
print("Name:", product['name'])
print("Price:", product['price'])
```

```python
# Question 2

product = {'name': 'pen', 'price': 20.0}

stock = 50
revised_price = 22.0

print("Before:", product)

product['stock'] = stock
product['price'] = revised_price

print("After:", product)
```

```python
# Question 3

employee = {'id': 'E101', 'name': 'Ravi'}

req_key = 'department'

print("Employee:", employee)

result = employee.get(req_key, "not available")

print("Requested Key:", req_key)
print("Result:", result)
```

```python
# Question 4

cart = {"Pen": 3, 'Notebook': 2, 'Marker': 1}
item_to_remove = 'Marker'

print("Before:", cart)

removed = cart.pop(item_to_remove)

print("Removed Quantity:", removed)
print("Remaining Cart:", cart)
```

```python
# Question 5

student = {'name': 'Asha', 'marks': {'Python': 82, 'Statistics': 76}}

marks = student["marks"] # {python:82, statistics:76}
python = marks["Python"]
stats = marks["Statistics"]

print(student)
print(marks)
print(python)
print(stats)
```

```python
# Question 6

defaults = {"theme": 'light', 'page_size':20}
user = {"theme": "dark"}

settings = defaults | user

print("Defaults", defaults)
print("User:", user)
print("Settings:", settings)
```

```python
# Question 8

original = {"theme": "light", "page_size": 20}

alias = original 
copied = original.copy()

print("Before:", original)

original["theme"] = "dark"

print("Original:", original)
print("Alias", alias)
print("Copied", copied)
```
