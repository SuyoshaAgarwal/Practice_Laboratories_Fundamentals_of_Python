```python
# Question 1

number = -7

if number < 0: 
    classification = "negative"
elif number == 0:
    classification = "zero"
else: 
    classification = "positive"

print(number)
print(classification)
```

```python
# Question 2

score = 84

is_valid = 0 <= score <= 100

if is_valid: 
    if score >= 90:
        grade = "A"
    elif score >= 80:
        grade = "B"
    elif score >= 70:
        grade = "C"
    elif score >= 60: 
        grade = "D"
    else: 
        grade = "F"
else: 
    grade = "Invalid Score"

print(score)
print(is_valid)
print(grade)
```

```python
# Question 3

year = 2028

div_by_400 = year % 400 == 0
div_by_4 = year % 4 == 0
div_by_100 = year % 100 == 0

is_leap_year = div_by_400 or (div_by_4 and not div_by_100)

print(year)
print(div_by_400)
print(div_by_4)
print(div_by_100)
print(is_leap_year)
```

```python
# Question 6

s1 = 5
s2 = 5
s3 = 8

is_valid = (s1 +s2 > s3 and s1 + s3 > s2 and s2 + s3 > s1)

if not is_valid: 
    triangle = "Invalid Triangle"
elif s1 == s2 == s3:
    triangle = "Equilateral Triangle"
elif s1 == s2 or s2 == s3 or s1 == s3:
    triangle = "Isosceles Triangle"
else:
    triangle = "Scalene Triangle"

print("Sides:", s1, s2, s3)
print("Valid:", is_valid)
print("Type:", triangle)
```

```python
# Question 7

day_num = 3

match day_num:
    case 1:
        name = "Monday"
    case 2:
        name = "Tuesday"
    case 3:
        name = "Wednesday"
    case 4:
        name = "Thursday"
    case 5: 
        name = "Friday"
    case 6: 
        name = "Saturday"
    case 7: 
        name = "Sunday"
    case 8: 
        name = "Invalid Day Number"

print("Day Number:", day_num)
print("Day Name:", name)
```

```python
# Question 8

transaction = ("withdrawal", 25000)

match transaction:
    case ("deposit", value):
        result = "Deposit accepted"
    case ("withdrawal", value) if value > 20000:
        result = "Withdrawal requires review"
    case ("withdrawal", value) if value > 0:
        result = "Withdrawal accepted"
    case _:
        result = "Unsupported transaction"
        
print(transaction)
print(result)
```
