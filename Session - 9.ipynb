```python
# Question 1

n = 10
total = 0

print("n:", n)

for number in range(1, n+1):
    total = total + number 
    print("Running total:", total)

print("Final Total:", total)
```

```python
# Question 2

number = 7

for multiplier in range (1, 11):
    product = number * multiplier
    print(number, "x", multiplier, "=", product)
```

```python
# Question 3

number = 6

factorial = 1

print("Number:", number)

for value in range(1, number + 1):
    factorial = factorial * value
    print("Running factorial:", factorial)

print("Final factorial:", factorial)
```

```python
# Question 4

number = 50842
working_number = number
digit_count = 0

print("Number:", number)

while working_number > 0:
    working_number = working_number // 10
    digit_count = digit_count + 1
    print("Working Number:", working_number)
    print("Digit Count:", digit_count)

print("Final Digit Count:", digit_count)
```

```python
# Question 5

values = [12, 7, 18, 5, 21]
target = 18
found_value = None

for value in values: 
    print("Checking:", value)

    if value == target:
        found_value = value 
        break
print("Target:", target)
print("Found:", found_value)
```

```python
# Question 6

entries = ["12", "7", "5", "done"]
total = 0

for entry in entries: 
    if entry == "done":
        break

    number = int(entry)
    total = total + number 
    print("Added:", number)
    print("Running Total:", total)

print("Final Total:", total )
```

```python
# Question 7

rows = 5

print("Rows:", rows)

for row in range(1, rows + 1):
    pattern = "*" * row
    print(pattern)
```

```python
# Question 8 

number = 29
is_prime = True

print("NUmber:", number)

for divisor in range(2, number):
    if number % divisor == 0:
        is_prime = False
        print("DIvisor found:", divisor)
        break 
print("Is prime:", is_prime)
```
