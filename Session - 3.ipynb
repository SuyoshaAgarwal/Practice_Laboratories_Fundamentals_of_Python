```python
# Question 1

text = "Data Science"

ltext = text.lower() # data science

vowels = (ltext.count("a") + 
ltext.count("i") + 
ltext.count("e") + 
ltext.count("o") + 
ltext.count("u"))

print("lowercase:", ltext)
print("Vowels:", vowels)
```

```python
# Question 2 

text = "Python"

reverse = text[::-1]

print("Original:", text)
print("Reversed:", reverse)
```

```python
# Question 3

text = "  Never Odd Or Even  "

normalised = text.strip().lower().replace(" ", "")

is_palin = normalised == normalised[::-1]

print(normalised)
print(is_palin)
```

```python
# Question 4

filename = "quarterly_report.csv"

parts = filename.split(".", 1) # [quaterly_report, csv]
extension = parts[1]

print("Parts:", parts)
print("Extension:", extension)
```

```python
# Question 5

text = "Python  makes  text  readable"

cleaned = text.replace("  ", " ")

print("Original:", text)
print("Cleaned:", cleaned)
```
